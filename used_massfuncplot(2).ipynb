```python
import numpy as np
import matplotlib.pyplot as plt

W = 100

vlist = np.linspace(0,400,1000)

def Massfunc(x):
    return (x**2)/(x**2+W**2)

Mlist = Massfunc(vlist)

plt.plot(vlist,Mlist)
plt.xlabel(r'$\tilde{v}$')
plt.ylabel(r'$M(\tilde{v})$')
plt.title(r'Mass function for $W = 100$')
plt.savefig('massfuncw100.pdf')
```
